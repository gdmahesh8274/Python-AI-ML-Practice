# Week 6 Day 4 - Unsupervised Learning Summary
Review clustering, PCA, and anomaly detection with small examples.

```python
import pandas as pd
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import IsolationForest

df = pd.DataFrame({'Hours': [1, 2, 3, 7, 8, 9, 25], 'Problems': [2, 3, 4, 8, 9, 10, 60]})
print(df)
```

```python
scaled = StandardScaler().fit_transform(df)
clusters = KMeans(n_clusters=2, n_init=10, random_state=42).fit_predict(scaled)
print('Cluster labels:', clusters)
```

```python
projection = PCA(n_components=2).fit_transform(scaled)
print('PCA projection:')
print(projection.round(2))
```

```python
anomalies = IsolationForest(contamination=1/7, random_state=42).fit_predict(scaled)
df['Possible_Anomaly'] = anomalies == -1
print(df)
```

## Notes
- K-Means groups similar rows without target labels.
- Scaling helps features contribute comparably to distance calculations.
- PCA creates new components that summarize variation in the data.
- Isolation Forest flags unusual rows for investigation; it does not prove an error.
- IQR is another simple method for detecting extreme values.
